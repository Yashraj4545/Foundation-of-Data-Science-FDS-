```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
df = pd.read_csv('student_dropout_behavior_dataset.csv')
```

```python
plt.figure(figsize=(8,5))
plt.plot(df['student_id'],df['lectures_attended'],marker="o",linewidth=2)
plt.title("Lecture Attendance Percentage by Student")
plt.xlabel("Student ID")
plt.ylabel("Attendance Percentage (%)")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
quiz_names = ["Quiz 1", "Quiz 2", "Quiz 3"]

quiz_averages = [
    df["quiz1_marks"].mean(),
    df["quiz2_marks"].mean(),
    df["quiz3_marks"].mean()
]

plt.figure(figsize=(8, 5))

plt.bar(quiz_names, quiz_averages)

plt.title("Average Quiz Marks")
plt.xlabel("Quiz")
plt.ylabel("Average Marks")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(
    df["final_marks"],
    bins=10,
    edgecolor="black"
)

plt.title("Distribution of Final Marks")
plt.xlabel("Final Marks")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["lectures_attended"],
    df["final_marks"],
    alpha=0.7
)

plt.title("Lecture Attendance vs Final Marks")
plt.xlabel("Lecture Attendance (%)")
plt.ylabel("Final Marks")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
male_marks = df[df["gender"] == "Male"]["final_marks"]
female_marks = df[df["gender"] == "Female"]["final_marks"]

plt.figure(figsize=(8, 5))

plt.boxplot(
    [male_marks, female_marks],
    labels=["Male", "Female"]
)

plt.title("Final Marks Distribution by Gender")
plt.xlabel("Gender")
plt.ylabel("Final Marks")

plt.tight_layout()
plt.show()
```

```python
gender_counts = df["gender"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    gender_counts.values,
    labels=gender_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Gender-wise Distribution of Students")

plt.tight_layout()
plt.show()
```

```python
numeric_columns = [
    "age",
    "quiz1_marks",
    "quiz2_marks",
    "quiz3_marks",
    "total_assignments",
    "midterm_marks",
    "final_marks",
    "previous_gpa",
    "total_lectures",
    "lectures_attended",
    "total_lab_sessions",
    "labs_attended"
]

correlation = df[numeric_columns].corr()

plt.figure(figsize=(12, 9))

plt.imshow(
    correlation,
    cmap="coolwarm",
    aspect="auto"
)

plt.colorbar(label="Correlation")

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=90
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

# Display correlation values
for i in range(len(correlation.columns)):
    for j in range(len(correlation.columns)):
        plt.text(
            j,
            i,
            f"{correlation.iloc[i, j]:.2f}",
            ha="center",
            va="center",
            fontsize=7
        )

plt.title("Correlation Heatmap of Student Performance")

plt.tight_layout()
plt.show()
```
